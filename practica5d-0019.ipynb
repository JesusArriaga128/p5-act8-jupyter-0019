```python
import pandas as pd
# Jesus Arriaga NC= 0019
# 1. Crear el diccionario con los datos
datos = {
    "Dia": ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"],
    "Temperatura": [12, 28, 33, 29, 27]
}

# 2. Convertir a DataFrame e imprimir
df = pd.DataFrame(datos)
print("=== TEMPERATURAS ===")
print(df)

# 3. Buscar la temperatura máxima y el día correspondiente
fila = df["Temperatura"].idxmax()

print("\nDía con mayor temperatura:", df.loc[fila, "Dia"])
print("Temperatura máxima:", df["Temperatura"].max(), "°C")
print("PROGRAMA REALIZADO POR JESUS ARRIAGA 0019")
```
