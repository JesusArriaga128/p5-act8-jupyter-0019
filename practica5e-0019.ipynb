```python
import pandas as pd
# Jesus Arriaga NC= 0019
# 1. Crear datos
datos = {
    "Producto": ["Escritorio", "Sillas", "Memoria ram", "Discos"],
    "Existencias": [8, 20, 4, 15]  # Se eliminó la palabra 'ado' de aquí
}

df = pd.DataFrame(datos)

# 2. Imprimir Inventario Completo
print("=== INVENTARIO ===")
print(df)

print("\n----------------------------------")

# 3. Filtrar Productos (< 10)
print("Productos con menos de 10 unidades:")
productos_bajos = df[df["Existencias"] < 10]
print(productos_bajos)

print("\n----------------------------------")

# 4. Suma Total
total = df["Existencias"].sum()
print("Existencias totales en almacén:", total)
print("PROGRAMA REALIZADO POR JESUS ARRIAGA 0019")
```
