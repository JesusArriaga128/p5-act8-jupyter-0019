```python
import pandas as pd
# Jesus Arriaga NC= 0019
# 1. Definición de datos
datos = {
    "Producto": ["hojas", "colores", "prgamento", "plumones", "globos"],
    "Precio": [35, 10, 250, 8, 15],
    "Cantidad": [5, 10, 2, 15, 6]
}

# 2. Creación del DataFrame
df = pd.DataFrame(datos)

# 3. Operación columna a columna
df["Total"] = df["Precio"] * df["Cantidad"]

# 4. Imprimir la tabla
print("=== VENTAS DE PAPELERÍA ===")
print(df)

# 5. Suma acumulada de la columna Total
print("\nVenta total:", df["Total"].sum())
print("PROGRAMA REALIZADO POR JESUS ARRIAGA 0019")
```
