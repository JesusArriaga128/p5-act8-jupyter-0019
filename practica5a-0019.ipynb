```python
import pandas as pd 
# practica calificaciones 
# Jesus Arriaga NC= 0019

datos = {
    "Nombre": ["Joel", "Mario", "Marísol", "Carla", "Sofía"],
    "Calificacion": [9, 8, 10, 7, 9]
}

# DataFrame: tabla o marco de datos 
df = pd.DataFrame(datos)

print("=== CALIFICACIONES NC= 0019 ===")
# Imprimir información de datos 
print(df)

print("\nPromedio y máximo:")
# Se cambia "calificacion" por "Calificacion" con C mayúscula
print("Promedio:", df["Calificacion"].mean())
print("Calificación mayor:", df["Calificacion"].max())

print("-+-+-+-+-+-+-+-jupyter Jesus-+-+-+-+-+-+-+-+-")
print("-+-+-+-+-+-+-+-jupyter 0019-+-+-+-+-+-+-+-+-")
print("PROGRAMA REALIZADO POR JESUS ARRIAGA 0019")
```
